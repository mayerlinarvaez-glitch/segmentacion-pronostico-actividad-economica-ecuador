INDICADOR DE TENDENCIA

```python
# Calcular la tendencia de las ventas por provincia

from scipy.stats import linregress

def calcular_tendencia(grupo):
    x = np.arange(len(grupo))
    y = grupo['TOTAL_VENTAS'].values

    pendiente = linregress(x, y).slope

    return pendiente

tendencia_provincia = (
    df_provincia_mes
    .sort_values(['PROVINCIA', 'PERIODO'])
    .groupby('PROVINCIA')
    .apply(calcular_tendencia)
    .rename('tendencia_ventas')
)

indicadores_provincia = indicadores_provincia.join(
    tendencia_provincia
)

indicadores_provincia
```

```python
# Crecimiento de las ventas entre 2020 y 2026

ventas_anuales = (
    df_provincia_mes
    .assign(AÑO=df_provincia_mes['PERIODO'].dt.year)
    .groupby(['PROVINCIA', 'AÑO'])['TOTAL_VENTAS']
    .sum()
    .reset_index()
)

ventas_2020 = (
    ventas_anuales[ventas_anuales['AÑO'] == 2020]
    .set_index('PROVINCIA')['TOTAL_VENTAS']
)

ventas_2026 = (
    ventas_anuales[ventas_anuales['AÑO'] == 2026]
    .set_index('PROVINCIA')['TOTAL_VENTAS']
)

crecimiento_2020_2026 = (
    (ventas_2026 / ventas_2020) - 1
).rename('crecimiento_2020_2026')

indicadores_provincia = indicadores_provincia.join(
    crecimiento_2020_2026
)

indicadores_provincia
```

```python

import numpy as np
import pandas as pd

# Copiar la base mensual provincial
serie = ventas_provincia_mes.copy()

serie["PERIODO"] = pd.to_datetime(serie["PERIODO"])
serie = serie.sort_values(["PROVINCIA", "PERIODO"])

# Calcular indicadores para cada provincia
def calcular_indicadores(grupo):
    grupo = grupo.sort_values("PERIODO").copy()
    ventas = grupo["TOTAL_VENTAS"].astype(float)

    # 1. Nivel: promedio mensual de ventas
    nivel = ventas.mean()

    # 2. Volatilidad: coeficiente de variación
    volatilidad = (
        ventas.std(ddof=1) / abs(nivel)
        if nivel != 0 else np.nan
    )

    # 3. Tendencia: pendiente lineal mensual
    tiempo = np.arange(len(ventas))
    tendencia = (
        np.polyfit(tiempo, ventas, 1)[0]
        if len(ventas) > 1 else np.nan
    )

    # 4. Estacionalidad: amplitud de promedios mensuales,
    # relativa al nivel promedio de ventas
    grupo["MES"] = grupo["PERIODO"].dt.month
    promedios_mes = grupo.groupby("MES")["TOTAL_VENTAS"].mean()
    estacionalidad = (
        (promedios_mes.max() - promedios_mes.min()) / abs(nivel)
        if nivel != 0 and len(promedios_mes) > 1 else np.nan
    )

    # 5. Choque pandémico: diferencia relativa entre 2020
    # y el período de referencia 2021-2025
    anios = grupo["PERIODO"].dt.year
    ventas_2020 = ventas[anios == 2020].mean()
    ventas_referencia = ventas[anios.between(2021, 2025)].mean()

    choque = (
        (ventas_2020 - ventas_referencia) / abs(ventas_referencia)
        if pd.notna(ventas_2020)
        and pd.notna(ventas_referencia)
        and ventas_referencia != 0
        else np.nan
    )

    return pd.Series({
        "nivel_ventas": nivel,
        "tendencia_ventas": tendencia,
        "estacionalidad": estacionalidad,
        "volatilidad": volatilidad,
        "choque_pandemico": choque
    })

indicadores_kmeans = (
    serie.groupby("PROVINCIA")
         .apply(calcular_indicadores)
)

print("Dimensiones:", indicadores_kmeans.shape)
print("\nValores nulos por indicador:")
display(indicadores_kmeans.isnull().sum())

print("\nIndicadores calculados:")
display(indicadores_kmeans.round(3))
```

```python

# ESTANDARIZACIÓN DE LOS INDICADORES PARA K-MEANS

from sklearn.preprocessing import StandardScaler
import pandas as pd

# Seleccionar las cinco variables definidas en los objetivos
variables_cluster = [
    "nivel_ventas",
    "tendencia_ventas",
    "estacionalidad",
    "volatilidad",
    "choque_pandemico"
]

# Preparar la matriz de indicadores
X_indicadores = indicadores_kmeans[variables_cluster].copy()

# Estandarizar las variables
escalador = StandardScaler()
X_kmeans = pd.DataFrame(
    escalador.fit_transform(X_indicadores),
    index=X_indicadores.index,
    columns=variables_cluster
)

print("Dimensiones de la matriz estandarizada:", X_kmeans.shape)

print("\nPromedio de cada variable estandarizada:")
display(X_kmeans.mean().round(3))

print("\nDesviación estándar de cada variable:")
display(X_kmeans.std(ddof=0).round(3))

print("\nPrimeras provincias:")
display(X_kmeans.head())
```

```python

# EVALUACIÓN DE K-MEANS: SILUETA Y DAVIES-BOULDIN

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score
import pandas as pd
import matplotlib.pyplot as plt

resultados_kmeans = []

for k in range(2, 8):
    modelo = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    etiquetas = modelo.fit_predict(X_kmeans)

    resultados_kmeans.append({
        "Clusters": k,
        "Silhouette": silhouette_score(X_kmeans, etiquetas),
        "Davies-Bouldin": davies_bouldin_score(X_kmeans, etiquetas),
        "Tamaño de grupos": sorted(
            pd.Series(etiquetas).value_counts().tolist()
        )
    })

tabla_kmeans = pd.DataFrame(resultados_kmeans)

display(tabla_kmeans.round(3))
```

```python

# Comparar la composición de K-means con 2, 3 y 4 clusters

from sklearn.cluster import KMeans
import pandas as pd

asignaciones_kmeans = indicadores_kmeans.copy()

for k in [2, 3, 4]:
    modelo = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    asignaciones_kmeans[f"cluster_{k}"] = (
        modelo.fit_predict(X_kmeans) + 1
    )

# Mostrar las provincias asignadas a cada grupo
for k in [2, 3, 4]:
    print(f"\n--- SOLUCIÓN DE {k} CLUSTERS ---")

    for grupo in sorted(asignaciones_kmeans[f"cluster_{k}"].unique()):
        provincias = asignaciones_kmeans.index[
            asignaciones_kmeans[f"cluster_{k}"] == grupo
        ].tolist()

        print(f"Cluster {grupo} ({len(provincias)} provincias):")
        print(", ".join(provincias))
```

```python

# PERFIL ECONÓMICO DE LOS CUATRO CLUSTERS DE K-MEANS

columnas_indicadores = [
    "nivel_ventas",
    "tendencia_ventas",
    "estacionalidad",
    "volatilidad",
    "choque_pandemico"
]

# Calcular el promedio de cada indicador por grupo
perfil_kmeans_4 = (
    asignaciones_kmeans
    .groupby("cluster_4")[columnas_indicadores]
    .mean()
)

print("Promedio de los indicadores por cluster:")
display(perfil_kmeans_4.round(3))

# Mostrar también la mediana, para reducir la influencia
# de provincias con valores extremos
print("\nMediana de los indicadores por cluster:")
display(
    asignaciones_kmeans
    .groupby("cluster_4")[columnas_indicadores]
    .median()
    .round(3)
)
```

```python

# Ranking de provincias según nivel y tendencia de ventas

columnas_ranking = [
    "nivel_ventas",
    "tendencia_ventas",
    "estacionalidad",
    "volatilidad",
    "choque_pandemico"
]

ranking_provincias = asignaciones_kmeans[
    columnas_ranking + ["cluster_4"]
].copy()

print("10 provincias con mayor nivel promedio de ventas:")
display(
    ranking_provincias
    .sort_values("nivel_ventas", ascending=False)
    .head(10)
    .round(3)
)

print("\n10 provincias con mayor tendencia positiva:")
display(
    ranking_provincias
    .sort_values("tendencia_ventas", ascending=False)
    .head(10)
    .round(3)
)
```

```python

# CLUSTERING CON DBSCAN

from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score, davies_bouldin_score
import pandas as pd
import numpy as np

resultados_dbscan = []

# Probar distintas combinaciones de eps y min_samples
for eps in [0.5, 0.8, 1.0, 1.2, 1.5, 2.0]:
    for min_samples in [2, 3, 4]:
        modelo_dbscan = DBSCAN(
            eps=eps,
            min_samples=min_samples
        )

        etiquetas = modelo_dbscan.fit_predict(X_kmeans)

        # Excluir ruido (-1) al evaluar los grupos
        mascara = etiquetas != -1
        etiquetas_validas = etiquetas[mascara]
        datos_validos = X_kmeans.loc[mascara]

        grupos = set(etiquetas_validas)
        n_grupos = len(grupos)
        n_ruido = int((etiquetas == -1).sum())

        silueta = np.nan
        davies = np.nan

        # Las métricas requieren al menos 2 grupos
        # y menos grupos que observaciones evaluadas
        if (
            n_grupos >= 2
            and len(etiquetas_validas) > n_grupos
        ):
            silueta = silhouette_score(
                datos_validos, etiquetas_validas
            )
            davies = davies_bouldin_score(
                datos_validos, etiquetas_validas
            )

        resultados_dbscan.append({
            "eps": eps,
            "min_samples": min_samples,
            "clusters_sin_ruido": n_grupos,
            "provincias_ruido": n_ruido,
            "Silhouette": silueta,
            "Davies-Bouldin": davies
        })

tabla_dbscan = pd.DataFrame(resultados_dbscan)

display(
    tabla_dbscan.sort_values(
        ["Silhouette", "Davies-Bouldin"],
        ascending=[False, True],
        na_position="last"
    ).round(3)
)
```

```python

import numpy as np
import pandas as pd
import warnings

from sklearn.metrics import mean_absolute_error, mean_squared_error
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.holtwinters import ExponentialSmoothing

warnings.filterwarnings("ignore")

# Guardar los resultados de todas las provincias
resultados_comparacion = []

# Guardar pronósticos para revisarlos posteriormente
pronosticos_provinciales = {}

# Recorrer las 24 provincias
provincias = sorted(base_pronostico["PROVINCIA"].unique())

for provincia in provincias:

    # Serie de entrenamiento
    train = (
        entrenamiento[
            entrenamiento["PROVINCIA"] == provincia
        ]
        .set_index("PERIODO")["TOTAL_VENTAS"]
        .sort_index()
        .asfreq("MS")
    )

    # Serie de prueba
    test = (
        prueba[
            prueba["PROVINCIA"] == provincia
        ]
        .set_index("PERIODO")["TOTAL_VENTAS"]
        .sort_index()
        .asfreq("MS")
    )

    # Comprobar que la serie esté completa
    if train.isna().any() or test.isna().any():
        print(f"Se omite {provincia}: hay valores faltantes.")
        continue

    # Comprobar valores finitos
    if not np.isfinite(train).all() or not np.isfinite(test).all():
        print(f"Se omite {provincia}: hay valores no finitos.")
        continue

    # Modelo 1: SARIMA estacional
    try:
        modelo_sarima = SARIMAX(
            train,
            order=(1, 0, 1),
            seasonal_order=(1, 0, 1, 12),
            enforce_stationarity=False,
            enforce_invertibility=False
        )

        ajuste_sarima = modelo_sarima.fit(disp=False)

        pred_sarima = ajuste_sarima.get_forecast(
            steps=len(test)
        ).predicted_mean

        pred_sarima.index = test.index

        mae = mean_absolute_error(test, pred_sarima)
        rmse = np.sqrt(mean_squared_error(test, pred_sarima))

        mascara = test != 0
        mape = (
            np.mean(
                np.abs(
                    (test[mascara] - pred_sarima[mascara])
                    / test[mascara]
                )
            ) * 100
            if mascara.any() else np.nan
        )

        resultados_comparacion.append({
            "PROVINCIA": provincia,
            "Modelo": "SARIMA",
            "MAE": mae,
            "RMSE": rmse,
            "MAPE": mape
        })

        pronosticos_provinciales[
            (provincia, "SARIMA")
        ] = pred_sarima

    except Exception as error:
        print(f"Error SARIMA en {provincia}: {error}")

    # Modelo 2: suavizamiento exponencial Holt-Winters
    try:
        modelo_hw = ExponentialSmoothing(
            train,
            trend="add",
            seasonal="add",
            seasonal_periods=12,
            initialization_method="estimated"
        )

        ajuste_hw = modelo_hw.fit(optimized=True)

        pred_hw = ajuste_hw.forecast(len(test))
        pred_hw.index = test.index

        mae = mean_absolute_error(test, pred_hw)
        rmse = np.sqrt(mean_squared_error(test, pred_hw))

        mascara = test != 0
        mape = (
            np.mean(
                np.abs(
                    (test[mascara] - pred_hw[mascara])
                    / test[mascara]
                )
            ) * 100
            if mascara.any() else np.nan
        )

        resultados_comparacion.append({
            "PROVINCIA": provincia,
            "Modelo": "Holt-Winters",
            "MAE": mae,
            "RMSE": rmse,
            "MAPE": mape
        })

        pronosticos_provinciales[
            (provincia, "Holt-Winters")
        ] = pred_hw

    except Exception as error:
        print(f"Error Holt-Winters en {provincia}: {error}")

# Crear la tabla de resultados
tabla_comparacion = pd.DataFrame(resultados_comparacion)

print("Provincias evaluadas:", tabla_comparacion["PROVINCIA"].nunique())
print("Modelos evaluados:", tabla_comparacion["Modelo"].value_counts())

display(
    tabla_comparacion
    .sort_values(["PROVINCIA", "RMSE"])
    .round(2)
    .reset_index(drop=True)
)
```

```python

# Diagnóstico de cobertura de las series provincia-sector

periodos_completos = pd.date_range(
    start="2020-01-01",
    end="2026-06-01",
    freq="MS"
)

diagnostico_series = []

for (provincia, sector), grupo in ventas_sectoriales.groupby(
    ["PROVINCIA", "CODIGO_SECTOR_N1"]
):
    serie = (
        grupo.set_index("PERIODO")["TOTAL_VENTAS"]
        .sort_index()
        .reindex(periodos_completos)
    )

    meses_faltantes = serie.isna().sum()
    meses_cero = (serie.fillna(np.nan) == 0).sum()
    meses_negativos = (serie.dropna() < 0).sum()

    periodos_observados = serie.dropna().index

    if len(periodos_observados) > 0:
        primer_mes = periodos_observados.min()
        ultimo_mes = periodos_observados.max()

        faltantes_iniciales = sum(
            periodo < primer_mes
            for periodo in periodos_completos
            if pd.isna(serie.loc[periodo])
        )

        faltantes_finales = sum(
            periodo > ultimo_mes
            for periodo in periodos_completos
            if pd.isna(serie.loc[periodo])
        )

        faltantes_internos = (
            meses_faltantes
            - faltantes_iniciales
            - faltantes_finales
        )
    else:
        primer_mes = pd.NaT
        ultimo_mes = pd.NaT
        faltantes_iniciales = len(periodos_completos)
        faltantes_finales = 0
        faltantes_internos = 0

    diagnostico_series.append({
        "PROVINCIA": provincia,
        "CODIGO_SECTOR_N1": sector,
        "meses_observados": serie.notna().sum(),
        "meses_faltantes": meses_faltantes,
        "faltantes_iniciales": faltantes_iniciales,
        "faltantes_internos": faltantes_internos,
        "faltantes_finales": faltantes_finales,
        "meses_con_ventas_cero": meses_cero,
        "meses_con_ventas_negativas": meses_negativos
    })

diagnostico_series = pd.DataFrame(diagnostico_series)

print("Resumen del diagnóstico:")
display(
    diagnostico_series[
        [
            "meses_observados",
            "meses_faltantes",
            "faltantes_iniciales",
            "faltantes_internos",
            "faltantes_finales",
            "meses_con_ventas_cero",
            "meses_con_ventas_negativas"
        ]
    ].describe().round(2)
)

print("\nSeries con meses faltantes internos:")
display(
    diagnostico_series[
        diagnostico_series["faltantes_internos"] > 0
    ].sort_values(
        ["faltantes_internos", "meses_observados"],
        ascending=[False, True]
    ).head(20)
)

print("\nSeries con ventas negativas:")
display(
    diagnostico_series[
        diagnostico_series["meses_con_ventas_negativas"] > 0
    ].head(20)
)
```

```python

# Preparar series sectoriales para la comparación de pronósticos

import pandas as pd
import numpy as np

sectores_prioritarios = ["G", "C", "A", "H", "M"]

fecha_corte = pd.Timestamp("2024-12-01")
fecha_final = pd.Timestamp("2026-06-01")

series_sectoriales_modelado = {}

for (provincia, sector), grupo in ventas_sectoriales_completas.groupby(
    ["PROVINCIA", "CODIGO_SECTOR_N1"]
):
    if sector not in sectores_prioritarios:
        continue

    grupo = grupo.sort_values("PERIODO")

    serie = (
        grupo.set_index("PERIODO")["TOTAL_VENTAS"]
        .asfreq("MS")
    )

    # Verificar que no existan meses ausentes
    if serie.isna().any():
        continue

    train = serie.loc[:fecha_corte]
    test = serie.loc[
        fecha_corte + pd.offsets.MonthBegin(1):fecha_final
    ]

    # Mantener únicamente series con 60 meses de entrenamiento
    # y 18 meses de prueba
    if len(train) == 60 and len(test) == 18:
        series_sectoriales_modelado[(provincia, sector)] = {
            "train": train,
            "test": test
        }

print(
    "Número de series provincia-sector preparadas:",
    len(series_sectoriales_modelado)
)

print(
    "Número de sectores representados:",
    len(set(sector for provincia, sector in series_sectoriales_modelado))
)

print(
    "Número de provincias representadas:",
    len(set(provincia for provincia, sector in series_sectoriales_modelado))
)

# Mostrar una serie de ejemplo
if series_sectoriales_modelado:
    clave_ejemplo = next(iter(series_sectoriales_modelado))
    ejemplo = series_sectoriales_modelado[clave_ejemplo]

    print("\nSerie de ejemplo:", clave_ejemplo)
    print("Entrenamiento:", len(ejemplo["train"]), "meses")
    print("Prueba:", len(ejemplo["test"]), "meses")
    print("Inicio del entrenamiento:", ejemplo["train"].index.min())
    print("Fin del entrenamiento:", ejemplo["train"].index.max())
    print("Inicio de la prueba:", ejemplo["test"].index.min())
    print("Fin de la prueba:", ejemplo["test"].index.max())
```

```python

import pandas as pd
import numpy as np
import warnings

from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from sklearn.metrics import mean_absolute_error, mean_squared_error

warnings.filterwarnings("ignore")

resultados_sectoriales = []
pronosticos_sectoriales = []

for (provincia, sector), datos in series_sectoriales_modelado.items():

    train = datos["train"].astype(float)
    test = datos["test"].astype(float)

    # Métricas y pronósticos de ambos modelos
    modelos = {
        "SARIMA": None,
        "Holt-Winters": None
    }

    # Modelo 1: SARIMA con estacionalidad anual
    try:
        modelo_sarima = SARIMAX(
            train,
            order=(1, 0, 1),
            seasonal_order=(1, 0, 1, 12),
            enforce_stationarity=False,
            enforce_invertibility=False
        )

        ajuste_sarima = modelo_sarima.fit(disp=False)
        pred_sarima = ajuste_sarima.forecast(steps=len(test))
        modelos["SARIMA"] = np.asarray(pred_sarima)

    except Exception as e:
        print(f"Error SARIMA en {provincia} - sector {sector}: {e}")

    # Modelo 2: Holt-Winters
    try:
        modelo_hw = ExponentialSmoothing(
            train,
            trend="add",
            seasonal="add",
            seasonal_periods=12,
            initialization_method="estimated"
        )

        ajuste_hw = modelo_hw.fit(optimized=True)
        pred_hw = ajuste_hw.forecast(len(test))
        modelos["Holt-Winters"] = np.asarray(pred_hw)

    except Exception as e:
        print(f"Error Holt-Winters en {provincia} - sector {sector}: {e}")

    # Evaluar cada modelo con los mismos datos de prueba
    for nombre_modelo, prediccion in modelos.items():

        if prediccion is None or len(prediccion) != len(test):
            continue

        reales = test.to_numpy()

        mae = mean_absolute_error(reales, prediccion)
        rmse = np.sqrt(mean_squared_error(reales, prediccion))

        # MAPE: excluir observaciones reales iguales a cero
        mascara = reales != 0

        if mascara.any():
            mape = np.mean(
                np.abs(
                    (reales[mascara] - prediccion[mascara])
                    / reales[mascara]
                )
            ) * 100
        else:
            mape = np.nan

        resultados_sectoriales.append({
            "PROVINCIA": provincia,
            "CODIGO_SECTOR_N1": sector,
            "Modelo": nombre_modelo,
            "MAE": mae,
            "RMSE": rmse,
            "MAPE": mape
        })

        for fecha, real, pronostico in zip(
            test.index, reales, prediccion
        ):
            pronosticos_sectoriales.append({
                "PROVINCIA": provincia,
                "CODIGO_SECTOR_N1": sector,
                "PERIODO": fecha,
                "Modelo": nombre_modelo,
                "Venta_real": real,
                "Pronostico": pronostico
            })

# Crear tablas finales
tabla_resultados_sectoriales = pd.DataFrame(resultados_sectoriales)
tabla_pronosticos_sectoriales = pd.DataFrame(pronosticos_sectoriales)

print("Evaluaciones completadas:", len(tabla_resultados_sectoriales))
print(
    "Series provincia-sector evaluadas:",
    tabla_resultados_sectoriales[
        ["PROVINCIA", "CODIGO_SECTOR_N1"]
    ].drop_duplicates().shape[0]
)

print("\nMÉTRICAS POR PROVINCIA Y SECTOR")
display(
    tabla_resultados_sectoriales
    .sort_values(["CODIGO_SECTOR_N1", "PROVINCIA", "MAPE"])
    .round(2)
    .reset_index(drop=True)
)

print("\nRESUMEN GLOBAL POR MODELO")
display(
    tabla_resultados_sectoriales
    .groupby("Modelo")
    .agg(
        series_evaluadas=("PROVINCIA", "count"),
        MAE_promedio=("MAE", "mean"),
        RMSE_promedio=("RMSE", "mean"),
        MAPE_promedio=("MAPE", "mean"),
        MAPE_mediano=("MAPE", "median")
    )
    .round(2)
)
```

```python

# Comparación de modelos por sector económico

resumen_sectorial = (
    tabla_resultados_sectoriales
    .groupby(["CODIGO_SECTOR_N1", "Modelo"])
    .agg(
        series_evaluadas=("PROVINCIA", "count"),
        MAE_promedio=("MAE", "mean"),
        RMSE_promedio=("RMSE", "mean"),
        MAPE_promedio=("MAPE", "mean"),
        MAPE_mediano=("MAPE", "median")
    )
    .round(2)
    .reset_index()
)

print("COMPARACIÓN DE MODELOS POR SECTOR")
display(resumen_sectorial)

# Identificar el modelo con menor error para cada sector
ganadores_sector = (
    resumen_sectorial
    .sort_values("MAPE_promedio")
    .groupby("CODIGO_SECTOR_N1")
    .first()
    .reset_index()
)

print("MODELO CON MENOR MAPE PROMEDIO POR SECTOR")
display(
    ganadores_sector[
        ["CODIGO_SECTOR_N1", "Modelo", "MAPE_promedio", "MAPE_mediano"]
    ]
)

# Contar en cuántos sectores gana cada modelo
print("NÚMERO DE SECTORES GANADOS")
display(
    ganadores_sector["Modelo"]
    .value_counts()
    .rename_axis("Modelo")
    .reset_index(name="Número de sectores")
)
```

```python

# Evaluación complementaria: WAPE por sector y modelo

evaluacion_wape = tabla_pronosticos_sectoriales.copy()

evaluacion_wape["Error_absoluto"] = (
    evaluacion_wape["Venta_real"] -
    evaluacion_wape["Pronostico"]
).abs()

resumen_wape = (
    evaluacion_wape
    .groupby(["CODIGO_SECTOR_N1", "Modelo"])
    .agg(
        Error_absoluto_total=("Error_absoluto", "sum"),
        Ventas_reales_absolutas=(
            "Venta_real",
            lambda x: x.abs().sum()
        ),
        MAE=("Error_absoluto", "mean"),
        RMSE=(
            "Error_absoluto",
            lambda x: np.sqrt(np.mean(x**2))
        )
    )
    .reset_index()
)

resumen_wape["WAPE"] = np.where(
    resumen_wape["Ventas_reales_absolutas"] > 0,
    resumen_wape["Error_absoluto_total"] /
    resumen_wape["Ventas_reales_absolutas"] * 100,
    np.nan
)

print("COMPARACIÓN COMPLEMENTARIA POR SECTOR")
display(
    resumen_wape[
        ["CODIGO_SECTOR_N1", "Modelo", "MAE", "RMSE", "WAPE"]
    ].round(2)
)

print("\nMODELO CON MENOR WAPE POR SECTOR")
ganadores_wape = (
    resumen_wape
    .sort_values("WAPE")
    .groupby("CODIGO_SECTOR_N1")
    .first()
    .reset_index()
)

display(
    ganadores_wape[
        ["CODIGO_SECTOR_N1", "Modelo", "WAPE"]
    ].round(2)
)
```

```python

# ============================================================
# PASO SIGUIENTE: SELECCIÓN DEL MEJOR MODELO POR PROVINCIA-SECTOR
# ============================================================

import pandas as pd
import numpy as np
from IPython.display import display

# 1. Copiar la tabla de predicciones mensuales ya calculadas
pronosticos_eval = tabla_pronosticos_sectoriales.copy()

# 2. Revisar y preparar los datos
pronosticos_eval["PERIODO"] = pd.to_datetime(
    pronosticos_eval["PERIODO"]
)

pronosticos_eval["Error_absoluto"] = (
    pronosticos_eval["Venta_real"] -
    pronosticos_eval["Pronostico"]
).abs()

# 3. Calcular métricas por provincia, sector y modelo
def calcular_metricas(grupo):
    ventas_reales = grupo["Venta_real"].abs().sum()
    suma_errores = grupo["Error_absoluto"].sum()

    wape = (
        suma_errores / ventas_reales * 100
        if ventas_reales != 0 else np.nan
    )

    return pd.Series({
        "Meses_evaluados": len(grupo),
        "MAE": grupo["Error_absoluto"].mean(),
        "RMSE": np.sqrt(
            np.mean(
                (grupo["Venta_real"] - grupo["Pronostico"]) ** 2
            )
        ),
        "MAPE": (
            np.mean(
                np.abs(
                    (grupo["Venta_real"] - grupo["Pronostico"])
                    / grupo["Venta_real"]
                )
            ) * 100
        ) if (grupo["Venta_real"] != 0).all() else np.nan,
        "WAPE": wape
    })

metricas_provincia_sector = (
    pronosticos_eval
    .groupby(
        ["PROVINCIA", "CODIGO_SECTOR_N1", "Modelo"]
    )
    .apply(calcular_metricas)
    .reset_index()
)

# 4. Comparar los modelos según el WAPE
comparacion_wape = metricas_provincia_sector.pivot(
    index=["PROVINCIA", "CODIGO_SECTOR_N1"],
    columns="Modelo",
    values="WAPE"
).reset_index()

# 5. Seleccionar el modelo con menor WAPE
modelos_disponibles = [
    modelo for modelo in ["SARIMA", "Holt-Winters"]
    if modelo in comparacion_wape.columns
]

comparacion_wape["Modelo_ganador_WAPE"] = (
    comparacion_wape[modelos_disponibles]
    .idxmin(axis=1)
)

comparacion_wape["WAPE_ganador"] = (
    comparacion_wape[modelos_disponibles]
    .min(axis=1)
)

# 6. Mostrar resultados
print("Series provincia-sector comparadas:",
      len(comparacion_wape))

print("\nModelos ganadores:")
display(
    comparacion_wape["Modelo_ganador_WAPE"]
    .value_counts()
    .rename_axis("Modelo")
    .to_frame("Número de series")
)

print("\nPrimeras 20 series seleccionadas:")
display(
    comparacion_wape
    .sort_values("WAPE_ganador")
    .head(20)
    .round(2)
)

# 7. Guardar la tabla consolidada para utilizarla después
comparacion_wape.to_csv(
    "seleccion_modelos_provincia_sector.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nTabla guardada como seleccion_modelos_provincia_sector.csv")
```

```python

# ============================================================
# PASO SIGUIENTE: RESUMEN DE GANADORES POR SECTOR ECONÓMICO
# ============================================================

# 1. Contar qué modelo gana en cada provincia-sector
ganadores_por_sector = (
    comparacion_wape
    .groupby(["CODIGO_SECTOR_N1", "Modelo_ganador_WAPE"])
    .size()
    .unstack(fill_value=0)
    .reset_index()
)

# 2. Calcular el porcentaje de provincias favorecidas
columnas_modelos = [
    col for col in ["Holt-Winters", "SARIMA"]
    if col in ganadores_por_sector.columns
]

ganadores_por_sector["Total_series"] = (
    ganadores_por_sector[columnas_modelos].sum(axis=1)
)

for modelo in columnas_modelos:
    ganadores_por_sector[f"Porcentaje_{modelo}"] = (
        ganadores_por_sector[modelo]
        / ganadores_por_sector["Total_series"] * 100
    ).round(2)

# 3. Calcular el WAPE promedio de cada modelo por sector
wape_promedio_sector = (
    metricas_provincia_sector
    .groupby(["CODIGO_SECTOR_N1", "Modelo"])["WAPE"]
    .mean()
    .unstack()
    .reset_index()
)

# 4. Unir ambos resúmenes
resumen_sectorial_modelos = ganadores_por_sector.merge(
    wape_promedio_sector,
    on="CODIGO_SECTOR_N1",
    how="left"
)

print("RESUMEN DE MODELOS GANADORES POR SECTOR")
display(resumen_sectorial_modelos.round(2))

print("\nWAPE promedio por sector y modelo")
display(wape_promedio_sector.round(2))
```
